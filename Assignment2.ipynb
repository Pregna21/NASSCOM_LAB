```python
import numpy as np
import pandas as pd
```

```python
#4. Part A – Python Programming
#creation of dataframe
data = {
    "Student":["S1","S2","S3","S4","S5","S6","S7","S8","S9","S10"],
    "Study_Hours":[2,3,4,5,6,7,8,4,6,9],
    "Attendance" :[65,70,75,80,85,90,92,78,88,95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94],
}
df = pd.DataFrame(data)
#2. Display the number of students.
print("\nNumber of students:", len(df))

# 3. Display column names
print("\nColumn names:", df.columns.tolist())

# 4. Display data types and non-null counts
print("\n--- Data Info ---")
df.info()

# 5. Display first 5 records
print("\n--- First 5 Records ---")
print(df.head())

# 6. Display last 5 records
print("\n--- Last 5 Records ---")
print(df.tail())
```

```python
#5. Part B – Pandas Data Analysis
#9. Calculate the average Study Hours, Attendance, Assignment Score, Previous Score and Final Score.
print("\n--- Average Study Hours:", df["Study_Hours"].mean())
print("\n--- Average Attendance:", df["Attendance"].mean())
print("\n--- Average Assignment Score:", df["Assignment_Score"].mean())
print("\n--- Average Previous Score:", df["Previous_Score"].mean())
print("\n--- Average Final Score:", df["Final_Score"].mean())
#10&11.Find Highest and Lowest Final Scores
print("\n--- Highest Final Score:", df["Final_Score"].max())
print("\n--- Lowest Final Score:", df["Final_Score"].min())
#12. Display students whose Attendance is greater than 80%.
print("\n--- Students with Attendance > 80%:")
print(df[df["Attendance"] > 80])
#13.Display students whose Final Score is greater than 80.
print("\n--- Students with Final Score > 80:")
print(df[df["Final_Score"] > 80])
#14.Use describe() to obtain a statistical summary.
print("\n--- Statistical Summary:")
print(df.describe())
```

```python
#6. Part C – NumPy Numerical Analysis
#Mean of final_score
print("Mean of final scores:", np.mean(df["Final_Score"]))
#Median of final_score
print("Median of final scores:", np.median(df["Final_Score"]))
#Max and Min of final_score
print("Max of final scores:", np.max(df["Final_Score"]))
print("Min of final scores:", np.min(df["Final_Score"]))
#Standard deviation of final_score
print("Standard deviation of final scores:", np.std(df["Final_Score"]))
```

```python
#7. Part D – Data Visualization
#Study Hours vs Final Score: Create a scatter plot. Comment on the visible relationship.
import matplotlib.pyplot as plt
plt.scatter(df["Study_Hours"], df["Final_Score"])
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.grid(True)
plt.show()
#Attendance vs Final Score: Create a scatter plot. Comment on the visible relationship.
plt.scatter(df["Attendance"], df["Final_Score"])
plt.xlabel("Attendance")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.grid(True)
plt.show()
# Distribution of Final Scores: Create a histogram. Comment on the distribution.
plt.hist(df["Final_Score"], bins=10)
plt.xlabel("Final Score")
plt.ylabel("Frequency")
plt.title("Distribution of Final Scores")
plt.grid(True)
plt.show()
#Student-wise Final Scores: Create a bar chart. Identify the highest and lowest scores.
plt.bar(df["Student"], df["Final_Score"])
plt.xlabel("Student")
plt.ylabel("Final Score")
plt.title("Student-wise Final Scores")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
#9. Part F – Calculus: Differentiation
def F(x):
    return 40 + 5 * x - 0.2 * (x**2)
def dF(x):
    return 5 - 0.4 * x
for x in range(1, 11):
    print(f"Hours: {x} | Score: {F(x):.1f} | Marginal Gain: {dF(x):.1f}")
```

```python
#10. Part G – Calculus: Integration
from scipy.integrate import quad

benefit, error = quad(lambda x: 5 * x - 0.2 * x**2, 0, 5)
print("Accumulated Learning Benefit:", benefit)
```

```python
#11. Part H – Correlation Analysis
df.corr(numeric_only=True)
import matplotlib.pyplot as plt
import seaborn as sns

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
#12. Part I – Mini Prediction Model Using Linear Algebra
import numpy as np

# Prepare features with intercept column (column of 1s)
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
X_design = np.column_stack((np.ones(len(X)), X))
Y = df["Final_Score"].values

# Solve for beta coefficients
beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y

print("Intercept (b0):", beta[0])
print("Study Hours Coeff (b1):", beta[1])
print("Attendance Coeff (b2):", beta[2])
print("Assignment Score Coeff (b3):", beta[3])
```
