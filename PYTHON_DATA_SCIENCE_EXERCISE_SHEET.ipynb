```python
#dataset and setup-----
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = {
"Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
111, 112, 113, 114, 115],
"Department": ["IT", "HR", "IT", "Finance", "HR",
"IT", "Finance", "IT", "HR", "Finance",
"IT", "HR", "Finance", "IT", "HR"],
"Age": [22, 25, 28, 30, 24, 35, 40, 27, 29, 45,
32, 26, 38, 31, 50],
"Experience": [1, 2, 4, 6, 2, 10, 15, 3, 5, 18,
8, 3, 12, 7, 20],
"Salary": [25000, 28000, 32000, 40000, 27000,
55000, 80000, 35000, 38000, 95000,
60000, 30000, 75000, 50000, 150000],
"Performance_Score": [65, 70, 75, 80, 68,
85, 90, 78, 72, 95,
88, 74, 92, 82, 96],
"Training_Completed": ["Yes", "No", "Yes", "Yes", "No",
"Yes", "Yes", "No", "Yes", "Yes",
"Yes", "No", "Yes", "Yes", "No"]
}
df = pd.DataFrame(data)
print(df)

#Display the first five records.
print("First five records:\n",df.head(5))
#Display the last five records.
print("Last five records:\n",df.tail(5))
#Find the number of rows and columns.
print("Number of rows and columns:\n",df.shape)
#Display the column names.
print("Column names:\n",df.columns)
#Display the data types.
print("Data types:\n",df.dtypes)


#Part B – Basic Data Exploration

#Generate descriptive statistics.
print("Descriptive statistics:\n",df.describe())
#Find the number of employees in each department.
print("Number of employees in each department:\n",df.groupby("Department").count())
#Find the number of employees who completed training.
print("Number of employees who completed training:\n",df[df["Training_Completed"]=="Yes"].count())
#Check whether there are missing values.
print("Missing values:\n",df.isnull().sum())

#Part C – Measures of Central Tendency

#Mean
print("Mean of Age:",df["Age"].mean())
print("Mean of Experience:",df["Experience"].mean())
print("Mean of Salary:",df["Salary"].mean())
print("Mean of Performance Score:",df["Performance_Score"].mean())
#Median
print("Median of Age:",df["Age"].median())
print("Median of Experience:",df["Experience"].median())
print("Median of Salary:",df["Salary"].median())
print("Median of Performance Score:",df["Performance_Score"].median())
#Mode
print("Mode of Age:\n",df["Age"].mode())
print("Mode of Experience:\n",df["Experience"].mode())
print("Mode of Salary:\n",df["Salary"].mode())
print("Mode of Performance_score:\n",df["Performance_Score"].mode())

#Part D – Measures of Spread / Dispersion
#Range
print("Range of Age:",df["Age"].max()-df["Age"].min())
#Variance
print("Variance of Age:",df["Age"].var())
print("Variance of Experience:",df["Experience"].var())
print("Variance of Salary:",df["Salary"].var())
print("Variance of Performance Score:",df["Performance_Score"].var())
#Standard Deviation
print("Standard Deviation of Age:",df["Age"].std())
print("Standard Deviation of Experience:",df["Experience"].std())
print("Standard Deviation of Salary:",df["Salary"].std())
print("Standard Deviation of Performance Score:",df["Performance_Score"].std())
#Quartiles and E. IQR
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)
IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

#Part E – Skewness
print("Skewness of Age:",df["Age"].skew())
print("Skewness of Experience:",df["Experience"].skew())
print("Skewness of Salary:",df["Salary"].skew())
print("Skewness of Performance Score:",df["Performance_Score"].skew())

#Part F – Visualization
#A. Histogram
plt.hist(df["Salary"], bins=6)
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()

#Part I – Group based analysis
print(df.groupby("Department")["Salary"].mean())
print(df.groupby("Department")["Salary"].median())
print(df.groupby("Department")["Salary"].std())

#Part J – Final Challenge
numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
  print("\n", column)
  print("Mean:", df[column].mean())
  print("Median:", df[column].median())
  print("Mode:", df[column].mode().tolist())
  print("Minimum:", df[column].min())
  print("Maximum:", df[column].max())
  print("Range:", df[column].max() - df[column].min())
  print("Variance:", df[column].var())
  print("Standard Deviation:", df[column].std())
  print("Q1:", df[column].quantile(0.25))
  print("Q3:", df[column].quantile(0.75))
  print("IQR:", df[column].quantile(0.75) -
  df[column].quantile(0.25))
  print("Skewness:", df[column].skew())
```
