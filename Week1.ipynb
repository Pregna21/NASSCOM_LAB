```python
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Pregna")
greet_user("Yoshitha")
```

```python
def greet_user(name="Guest"):
  print(f"Hello,{name}!!")

greet_user()
greet_user("Pregna")
```

```python
def greet_user(*names):
  for name in names:
    print(f"Hello ,{name}!!")

greet_user("pregna","divya","xyz")
```

```python
def greet_user(**details):
  for key,value in details.items():
    print(f"{key}={value}")

greet_user(Name="Pregna",Age=17,College="BVRITH")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
def multiply(a):
  return a*2
x=map(multiply,(1,2,3))
print(list(x))
```

```python
numbers=[1,2,3,4]
even_numbers=list(filter(lambda x:x%2==0,numbers))
square=list(map(lambda x:x*x,even_numbers))
print(square)
```

```python
from functools import reduce
numbers=[1,2,3,4,5]
reduced=reduce(lambda x,y:x+y,numbers)
print(reduced)
```

```python
info = {"marks1": 92, "marks2": 72, "marks3": 45}
new_info=dict(map(lambda item:(item[0],item[1]+3),info.items()))

print(new_info)
```

```python
info = {"marks1": 92, "marks2": 72, "marks3": 45}
passed=dict(filter(lambda item:item[1]>50,info.items()))
print(passed)
```

```python
from functools import reduce
info = {"marks1": 92, "marks2": 72, "marks3": 45}
total=reduce(lambda a,b:a+b,info.values())
print(total)
```
